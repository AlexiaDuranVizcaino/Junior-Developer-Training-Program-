# freeCodeCamp — JavaScript Basics

## JS Basics — Cheat Sheet

- **Comments:** `// inline` · `/* multi-line */`
- **8 data types:** undefined, null, boolean, string, symbol, bigint, number, object
- **Declare:** `var name;` (ends with `;`; no spaces, no leading number)
- **Assign `=`:** right side runs first → stored on left
- **Init:** `var x = 9;` (declare + assign in one line)
- **String:** chars in `" "` or `' '`
- **Uninitialized** → `undefined`; math on it → `NaN`; with string → `"undefined"`

````python
var a = 7;            // declare + init
var b = a;            // copy value -> 7
var name = "Alexia";  // string

var c = 5;
c = c + 1;            // 6  (init avoids undefined/NaN)
````

## Variables: casing, let & const

- **Case sensitive:** `MYVAR` ≠ `MyVar` ≠ `myvar`. Use **camelCase** (`someVariableName`).
- **`var`:** can be re-declared/overwritten → risky bugs (no error thrown).
- **`let`:** same name can only be declared **once** in scope.
- **`const`:** read-only, cannot be reassigned. Use for values that never change; UPPERCASE is a common convention for immutables.
- Rule of thumb: `const` by default, `let` when it must change, avoid `var`.


## Numbers & Operators

- **Arithmetic:** `+` add · `-` subtract · `*` multiply · `/` divide.
- **Decimals (floats):** e.g. `5.7`; floating-point math has finite precision.
- **Remainder `%`:** gives leftover of a division (`11 % 3` → `2`); even/odd check with `n % 2`.
- **Increment / decrement:** `i++` (= `i = i + 1`), `i--` (= `i = i - 1`).
- **Compound assignment:** `+=`, `-=`, `*=`, `/=` combine an operation with assignment (`x += 5`).


## Strings

**Escaping & quotes**
- Start/end with same quote (`"` or `'`). Use one type to include the other without escaping: `'He said, "Hi"'`.
- **Escape a quote:** `\"` or `\'` so it appears inside the string.
- **Escape sequences:** `\'` `\"` `\\` (backslash) · `\n` newline · `\t` tab · `\r` carriage return · `\b` backspace · `\f` form feed. (`\` ≠ `/`)

**Concatenation**
- **`+`:** join strings → `"a" + "b"`; spaces are NOT added automatically.
- **`+=`:** append onto an existing string → `str += "more"`.
- **Insert variables:** `"Hello " + name + "!"` or build up with `+=` (word blanks).

**Length & bracket notation**
- **`.length`:** number of characters, spaces included → `"Alan Peter".length` = `10`.
- **Zero-based indexing:** first char `str[0]`, Nth char `str[1]`, `str[2]`...
- **Last char:** `str[str.length - 1]`; Nth-to-last: `str[str.length - N]`.
- **Immutable:** can't change a char (`str[0] = "J"` fails); only reassign the whole string.


## Arrays

- **Declare:** `const arr = ["a", 1, true];` — can mix types.
- **Nested / multi-dimensional:** `[["Bulls", 23], ["Sox", 45]]`.
- **Access by index (0-based):** `arr[0]`; multi-dim: `arr[3][0][1]`.
- **Mutable even with `const`:** change items → `arr[0] = 45` (contents can change; the binding can't be reassigned).
- **Style:** no space between name and brackets (`arr[0]`, not `arr [0]`).

**Add / remove elements**
- **`push(x)`:** add to the **end**, returns new length.
- **`pop()`:** remove **last**, returns it.
- **`shift()`:** remove **first**, returns it.
- **`unshift(x)`:** add to the **beginning**.


## Functions & Scope

**Basics**
- **Define:** `function name() { ... }`; **call:** `name();`.
- **Parameters vs arguments:** params are placeholders in the definition; args are the real values passed on call → `function add(a, b) {}` called as `add(1, 2)`.
- **`return`:** sends a value out of the function → `const r = timesFive(5);`. No `return` → function returns `undefined`.
- **Assign a return value:** right side runs first → `processed = processArg(7);`.

**Scope**
- **Global:** declared outside any function → visible everywhere.
- **Local:** declared inside a function (and params) → visible only there.
- **Same name:** local wins (shadows the global) inside the function.
- Always use `let`/`const`; a variable assigned without them becomes an accidental global.


## Queues, Booleans & if statements

- **Queue (FIFO):** add at back, remove from front → `push()` + `shift()`.
- **Booleans:** only `true` / `false` (no quotes).
- **`if`:** runs block only when condition is `true`.

```js
function trueOrFalse(x) {
  if (x) {
    return "Yes";
  }
  return "No";
}
```
